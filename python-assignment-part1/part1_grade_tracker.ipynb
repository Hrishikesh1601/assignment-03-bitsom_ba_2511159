## Task 1 Data Parsing & Cleaning

```python

raw_students = [
    {"name": "  ayesha SHARMA  ", "roll": "101", "marks_str": "88, 72, 95, 60, 78"},
    {"name": "ROHIT verma", "roll": "102", "marks_str": "55, 68, 49, 72, 61"},
    {"name": "  Priya Nair  ", "roll": "103", "marks_str": "91, 85, 88, 94, 79"},
    {"name": "karan MEHTA", "roll": "104", "marks_str": "40, 55, 38, 62, 50"},
    {"name": " Sneha pillai ", "roll": "105", "marks_str": "75, 80, 70, 68, 85"},
]

print("Task 1 Output:\n")

for student in raw_students:
    
    # fixing name step by step
    name = student["name"]
    name = name.strip()
    name = name.title()
    
    # converting roll
    roll = int(student["roll"])
    
    # converting marks string into list
    marks_str = student["marks_str"]
    parts = marks_str.split(",")
    
    marks = []
    for p in parts:
        marks.append(int(p.strip()))
    
    # checking if name is valid
    valid = True
    words = name.split()
    
    for w in words:
        if w.isalpha() == False:
            valid = False
    
    if valid:
        print("✓ Valid name")
    else:
        print("✗ Invalid name")
    
    # printing student profile
    print("="*32)
    print("Student :", name)
    print("Roll No :", roll)
    print("Marks   :", marks)
    print("="*32)
    
    # special case
    if roll == 103:
        print(name.upper())
        print(name.lower())
    
    print()
    
#Summary: 
# Here Student data is not clean:
# 1.Names have extra spaces and random casing.
# 2.Roll numbers are strings not numbers.
# 3.Marks are in one string instead of list.
```

## TAsk 2 Marks Analysis

```python
print("\nTask 2 Output:\n")

student_name = "Ayesha Sharma"
subjects = ["Math", "Physics", "CS", "English", "Chemistry"]
marks = [88, 72, 95, 60, 78]

# printing subjects with grades
for i in range(len(subjects)):
    
    m = marks[i]
    
    if m >= 90:
        grade = "A+"
    elif m >= 80:
        grade = "A"
    elif m >= 70:
        grade = "B"
    elif m >= 60:
        grade = "C"
    else:
        grade = "F"
    
    print(subjects[i], ":", m, "-", grade)

# calculating total and average
total = sum(marks)
avg = round(total / len(marks), 2)

print("\nTotal =", total)
print("Average =", avg)

# highest marks
max_marks = max(marks)
max_index = marks.index(max_marks)
print("Highest subject:", subjects[max_index], "-", max_marks)

# lowest marks
min_marks = min(marks)
min_index = marks.index(min_marks)
print("Lowest subject:", subjects[min_index], "-", min_marks)


# adding new subjects using while loop
count = 0

while True:
    sub = input("Enter subject (or done): ")
    
    if sub.lower() == "done":
        break
    
    val = input("Enter marks: ")
    
    # checking valid number
    if not val.isdigit():
        print("Invalid input, try again")
        continue
    
    val = int(val)
    
    if val < 0 or val > 100:
        print("Marks should be between 0 and 100")
        continue
    
    subjects.append(sub)
    marks.append(val)
    count += 1

print("\nNew subjects added:", count)
print("Updated average:", round(sum(marks)/len(marks), 2))

# Summary:
# 1.Assigned grades using conditions.
# 2.Calculated total & Average.
# 3.Found highest and lowest
# 4.Input Validation
# 5.Updating data dynamically.
```

## Task 3 Class Summary

```python
print("\nTask 3 Output:\n")

class_data = [
    ("Ayesha Sharma",  [88, 72, 95, 60, 78]),
    ("Rohit Verma",    [55, 68, 49, 72, 61]),
    ("Priya Nair",     [91, 85, 88, 94, 79]),
    ("Karan Mehta",    [40, 55, 38, 62, 50]),
    ("Sneha Pillai",   [75, 80, 70, 68, 85]),
]

print("Name              | Average | Status")
print("----------------------------------------")

pass_count = 0
fail_count = 0
total_avg = 0

top_name = ""
top_avg = 0

for data in class_data:
    
    name = data[0]
    marks = data[1]
    
    avg = round(sum(marks)/len(marks), 2)
    total_avg += avg
    
    if avg >= 60:
        status = "Pass"
        pass_count += 1
    else:
        status = "Fail"
        fail_count += 1
    
    print(f"{name:<18} | {avg:^7} | {status}")
    
    # checking topper
    if avg > top_avg:
        top_avg = avg
        top_name = name

print("\nPassed:", pass_count)
print("Failed:", fail_count)
print("Topper:", top_name, "-", top_avg)
print("Class Average:", round(total_avg/len(class_data), 2))


#Summary:
# 1.Calculated average per student.
# 2.Assigned,Counted Pass/Fail
# 3.Calculated class average.
```

## Task 4 String Operations

```python
print("\nTask 4 Output:\n")

essay = "  python is a versatile language. it supports object oriented, functional, and procedural programming. python is widely used in data science and machine learning.  "

# step 1 - cleaning spaces
clean_essay = essay.strip()
print("Cleaned essay:")
print(clean_essay)

# step 2 - title case
print("\nTitle Case:")
print(clean_essay.title())

# step 3 - count word
count = clean_essay.count("python")
print("\n'python' count:", count)

# step 4 - replace
replaced = clean_essay.replace("python", "Python 🐍")
print("\nAfter replace:")
print(replaced)

# step 5 - split sentences
sentences = clean_essay.split(". ")
print("\nSentence list:")
print(sentences)

# step 6 - numbered sentences
print("\nNumbered sentences:")

i = 1
for s in sentences:
    if s.endswith(".") == False:
        s = s + "."
    print(str(i) + ".", s)
    i += 1

    
# Summary:

# 1.Counted word occurence
# 2.Replaced words,Spilt into sentences,Numbered sentences.
```
