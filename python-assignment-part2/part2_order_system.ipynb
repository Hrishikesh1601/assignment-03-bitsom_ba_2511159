## Task 1 Explore the Menu

```python
import copy   # for deep copy

menu = {
    "Paneer Tikka":   {"category": "Starters",  "price": 180.0, "available": True},
    "Chicken Wings":  {"category": "Starters",  "price": 220.0, "available": False},
    "Veg Soup":       {"category": "Starters",  "price": 120.0, "available": True},
    "Butter Chicken": {"category": "Mains",     "price": 320.0, "available": True},
    "Dal Tadka":      {"category": "Mains",     "price": 180.0, "available": True},
    "Veg Biryani":    {"category": "Mains",     "price": 250.0, "available": True},
    "Garlic Naan":    {"category": "Mains",     "price":  40.0, "available": True},
    "Gulab Jamun":    {"category": "Desserts",  "price":  90.0, "available": True},
    "Rasgulla":       {"category": "Desserts",  "price":  80.0, "available": True},
    "Ice Cream":      {"category": "Desserts",  "price": 110.0, "available": False},
}

inventory = {
    "Paneer Tikka":   {"stock": 10, "reorder_level": 3},
    "Chicken Wings":  {"stock":  8, "reorder_level": 2},
    "Veg Soup":       {"stock": 15, "reorder_level": 5},
    "Butter Chicken": {"stock": 12, "reorder_level": 4},
    "Dal Tadka":      {"stock": 20, "reorder_level": 5},
    "Veg Biryani":    {"stock":  6, "reorder_level": 3},
    "Garlic Naan":    {"stock": 30, "reorder_level": 10},
    "Gulab Jamun":    {"stock":  5, "reorder_level": 2},
    "Rasgulla":       {"stock":  4, "reorder_level": 3},
    "Ice Cream":      {"stock":  7, "reorder_level": 4},
}

sales_log = {
    "2025-01-01": [
        {"order_id": 1,  "items": ["Paneer Tikka", "Garlic Naan"], "total": 220.0},
        {"order_id": 2,  "items": ["Gulab Jamun", "Veg Soup"], "total": 210.0},
        {"order_id": 3,  "items": ["Butter Chicken", "Garlic Naan"], "total": 360.0},
    ],
    "2025-01-02": [
        {"order_id": 4,  "items": ["Dal Tadka", "Garlic Naan"], "total": 220.0},
        {"order_id": 5,  "items": ["Veg Biryani", "Gulab Jamun"], "total": 340.0},
    ],
    "2025-01-03": [
        {"order_id": 6,  "items": ["Paneer Tikka", "Rasgulla"], "total": 260.0},
        {"order_id": 7,  "items": ["Butter Chicken", "Veg Biryani"], "total": 570.0},
        {"order_id": 8,  "items": ["Garlic Naan", "Gulab Jamun"], "total": 130.0},
    ],
    "2025-01-04": [
        {"order_id": 9,  "items": ["Dal Tadka", "Garlic Naan", "Rasgulla"], "total": 300.0},
        {"order_id": 10, "items": ["Paneer Tikka", "Gulab Jamun"], "total": 270.0},
    ],
}

# TASK 1 - EXPLORE THE MENU

print("\n")

categories = ["Starters", "Mains", "Desserts"]

for cat in categories:
    print(f"\n===== {cat} =====")
    for item in menu:
        if menu[item]["category"] == cat:
            status = "Available" if menu[item]["available"] else "Unavailable"
            print(f"{item} - ₹{menu[item]['price']} [{status}]")

# simple calculations
total_items = len(menu)
available_items = 0
max_price = 0
expensive_item = ""

cheap_items = []

for item in menu:
    if menu[item]["available"]:
        available_items += 1

    if menu[item]["price"] > max_price:
        max_price = menu[item]["price"]
        expensive_item = item

    if menu[item]["price"] < 150:
        cheap_items.append(item)

print("\nTotal items:", total_items)
print("Available items:", available_items)
print("Most expensive:", expensive_item, "-", max_price)

print("Items under 150:")
for i in cheap_items:
    print(i, "-", menu[i]["price"])

# Summary:
# 1.Printing Menu by Category.
# 2.Calculations: Total items,Available items,Most Expensive item,items under 150.
```

## Task 2 Cart Operations

```python
# TASK 2 - CART OPERATIONS

cart = []

def add_item(name, qty):
    if name not in menu:
        print("Item not found:", name)
        return
    
    if not menu[name]["available"]:
        print("Item not available:", name)
        return
    
    # check if already in cart
    for c in cart:
        if c["item"] == name:
            c["quantity"] += qty
            print(name, "quantity updated")
            return
    
    cart.append({
        "item": name,
        "quantity": qty,
        "price": menu[name]["price"]
    })
    print(name, "added to cart")


def remove_item(name):
    for c in cart:
        if c["item"] == name:
            cart.remove(c)
            print(name, "removed")
            return
    print("Item not in cart:", name)


# simulate steps
add_item("Paneer Tikka", 2)
add_item("Gulab Jamun", 1)
add_item("Paneer Tikka", 1)
add_item("Mystery Burger", 1)
add_item("Chicken Wings", 1)

remove_item("Gulab Jamun")

# order summary
print("\n----- BILL -----")

subtotal = 0

for c in cart:
    total = c["quantity"] * c["price"]
    subtotal += total
    print(c["item"], "x", c["quantity"], "=", total)

gst = subtotal * 0.05
final = subtotal + gst

print("Subtotal:", subtotal)
print("GST:", round(gst, 2))
print("Total:", round(final, 2))

# Summary:
# 1.Cart Structure
# 2.Add item function
# 3.Remove item function
# 4.Simulation
# 5.Bill calculation
```

## Task 3 Inventory Tracker with Deep Copy

```python
# TASK 3 - INVENTORY TRACKER WITH DEEP COPY

inventory_backup = copy.deepcopy(inventory)

# test deep copy
inventory["Paneer Tikka"]["stock"] = 1

print("\nCheck deep copy:")
print("Current:", inventory["Paneer Tikka"]["stock"])
print("Backup:", inventory_backup["Paneer Tikka"]["stock"])

# restore
inventory = copy.deepcopy(inventory_backup)

# deduct stock
for c in cart:
    item = c["item"]
    qty = c["quantity"]

    if inventory[item]["stock"] >= qty:
        inventory[item]["stock"] -= qty
    else:
        print("Not enough stock for", item)
        inventory[item]["stock"] = 0

# reorder alert
print("\nReorder Alerts:")
for item in inventory:
    if inventory[item]["stock"] <= inventory[item]["reorder_level"]:
        print(item, "low stock:", inventory[item]["stock"])

# Summary:
# 1.Deep Copy
# 2.Restore inventory
# 3.Deduct stock
# 4.Reorder alert
```

## Task 4 Daily Sales Log Analysis

```python
# TASK 4 - DAILY SALES LOG ANALYSIS

print("\nDaily Revenue:")

for date in sales_log:
    total = 0
    for order in sales_log[date]:
        total += order["total"]
    print(date, "=", total)

# best day
best_day = ""
best_value = 0

for date in sales_log:
    total = sum(o["total"] for o in sales_log[date])
    if total > best_value:
        best_value = total
        best_day = date

print("Best day:", best_day)

# most ordered item
count = {}

for date in sales_log:
    for order in sales_log[date]:
        for item in order["items"]:
            count[item] = count.get(item, 0) + 1

print("Most ordered item:", max(count, key=count.get))

# add new day
sales_log["2025-01-05"] = [
    {"order_id": 11, "items": ["Butter Chicken", "Gulab Jamun", "Garlic Naan"], "total": 490.0},
    {"order_id": 12, "items": ["Paneer Tikka", "Rasgulla"], "total": 260.0},
]

# enumerate orders
print("\nAll Orders:")

num = 1
for date in sales_log:
    for order in sales_log[date]:
        print(num, date, order["order_id"], order["total"], order["items"])
        num += 1

# Summary:
# 1.Daily revenue
# 2.Best Selling Day
# 3.Most ordered item
# 4.Enumerate orders
```
